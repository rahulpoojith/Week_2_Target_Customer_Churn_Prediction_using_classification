# 1. Download and Clean the Data

Run the notebooks in order: this one first, then `02_eda_feature_engineering.ipynb`, then `03_model_building.ipynb`. This notebook downloads the data, removes the customer ID, and saves a clean copy.

```python
import os
import shutil
from pathlib import Path

import kagglehub
import pandas as pd
from dotenv import load_dotenv
```

```python
PROJECT_DIR = Path.cwd().parent
load_dotenv(PROJECT_DIR / ".env")

raw_folder = PROJECT_DIR / "Datasets"
clean_folder = PROJECT_DIR / "Cleaned_Datasets"
raw_folder.mkdir(exist_ok=True)
clean_folder.mkdir(exist_ok=True)

dataset_name = os.getenv("KAGGLE_DATASET", "blastchar/telco-customer-churn")
download_folder = Path(kagglehub.dataset_download(dataset_name))
csv_file = list(download_folder.rglob("*.csv"))[0]

raw_file = raw_folder / "telco_customer_churn.csv"
shutil.copyfile(csv_file, raw_file)
data = pd.read_csv(raw_file)

print("Rows and columns:", data.shape)
display(data.head())
```

## Clean and save

```python
data = data.drop_duplicates()
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"], errors="coerce")
data["Churn"] = data["Churn"].map({"Yes": 1, "No": 0})
data = data.drop(columns="customerID")

clean_file = clean_folder / "telco_customer_churn.csv"
data.to_csv(clean_file, index=False)

print("Saved cleaned data:", data.shape)
print("Missing TotalCharges values:", data["TotalCharges"].isna().sum())
display(data.head())
```
